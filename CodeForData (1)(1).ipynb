```python
import pandas as pd
import matplotlib.pyplot as plt
publicSaftey = pd.read_csv("PublicSaftey.csv")
publicSaftey = publicSaftey.sort_values(by="Population (2010)", ascending=True)
```

```python
publicSaftey.iloc[80:90].plot(
    x='Neighborhood',
    y='Population (2010)',
    kind='bar',
    figsize=(10,6),
    color='blue',
    title='Population by Neighborhood (2010)'
)

plt.xlabel('Neighborhood')
plt.ylabel('Population (2010)')
plt.tight_layout()
plt.show()
```

```python
popDensity = pd.read_csv("population-density.csv")
popDensity['Land Area (acres)'] = (
    popDensity['Land Area (acres)']
    .astype(str)
    .str.replace(',', '')
    .astype(float)
)
popDensity = popDensity.sort_values(by="Land Area (acres)", ascending=True)
popDensity.iloc[79:89].plot(
    x='Neighborhood',
    y='Land Area (acres)',
    kind='bar',
    figsize=(10,6),
    color='pink',
    title='Land Area (acres) for Neighborhoods'
)

plt.xlabel('Neighborhood')
plt.ylabel('Land Area (acres)')
plt.tight_layout()
plt.show()
```

As you can see, the size of Brookline is very fitting for its population. Since it does have a large population the only way for that to be a bad thing is for it to not be big enough for to support that. But this is not the case for Brookline.
