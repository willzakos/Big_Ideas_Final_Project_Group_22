For my individual metric, I analyzed educational enrollment across Pittsburgh neighborhoods using the WPRDC dataset “School Enrollment by Detailed Level of School.” The two sub datasets I chose was kindergarten enrollment and grade 12 enrollment because they measure both early and late stages of the K–12 education system. This factor shows that those neighborhoods with high student enrollment most likely have a strong family community, which neighborhoods retain students through the end of high school, as well and more accessible schools. My goal was to use these indicators to determine which neighborhood performs the best according to the education-based metric.

```python
import pandas as pd
df = pd.read_csv("https://data.wprdc.org/dataset/3cb2b9c2-66e2-4ffa-b0a1-35fb2c6f7230/resource/eb1f8714-e0eb-4614-acfe-dfe0ebab942f/download/school-enrollment-by-detailed-level-of-school-for-the-population-3-years-and-over.csv")
kindergarten_ranked = df[["Neighborhood","Estimate; Enrolled in school: - Enrolled in kindergarten"]].sort_values(
    by="Estimate; Enrolled in school: - Enrolled in kindergarten",
    ascending = False
)
kindergarten_ranked.head(25)

kindergarten_ranked = kindergarten_ranked.set_index("Neighborhood")

print(kindergarten_ranked.head(25).to_string())

```

1. Loading the Dataset:
I used pandas.read_csv() to import the WPRDC dataset into a DataFrame. Then I selected only the two columns I needed:
"Estimate; Enrolled in school: - Enrolled in kindergarten"
"Estimate; Enrolled in school: - Enrolled in grade 12"
These columns already contain numerical enrollment counts for every neighborhood.

2. Ranking Neighborhoods:
I used .sort_values() to sort the neighborhoods from highest to lowest enrollment for each grade level. This gave me a ranked list showing the neighborhoods with the most kindergarten students. 

```python
kindergarten_ranked = df[["Neighborhood","Estimate; Enrolled in school: - Enrolled in kindergarten"]].sort_values(
    by="Estimate; Enrolled in school: - Enrolled in kindergarten",
    ascending = False
)
kindergarten_ranked.head(25)

kindergarten_ranked = kindergarten_ranked.set_index("Neighborhood")

kindergarten_ranked.head(25).plot(
    kind="bar",
    legend=False,
    figsize=(10,5),
    title="Top 25 Pittsburgh Neighborhoods by Kindergarten Enrollment"
)
```

3. Bar Graph:
I used the .plot(kind="bar") function on the top 25 neighborhoods in each category:
- produced a clear bar graph showing how the top neighborhoods compare

The graph here shows the amount of Kindergarten students enrolled for each neighborhood. Here, Brookline is ranked #2, with an estimate of 125 students enrolled from there. This shows that many young families choose to live here, which also relates to the fact young families choose to live here because of the fact that Brookline is relatively safe. 


```python
grade_12_ranked = df[["Neighborhood","Estimate; Enrolled in school: - Enrolled in grade 12"]].sort_values(
    by="Estimate; Enrolled in school: - Enrolled in grade 12",
    ascending = False
)
grade_12_ranked.head(25)

grade_12_ranked = grade_12_ranked.set_index("Neighborhood")

print(grade_12_ranked.head(25).to_string())

```

Here, I did followed the same line of code as the Kindergarten Enrollment dataset but with just a different column name. Based off this list, you can see that Brookline is still ranked #4 accross neighborhoods for most students enrolled in the 12th grade, making it one of the only neighborhoods with a consistent high ranked enrollment. 

```python
grade_12_ranked = df[["Neighborhood","Estimate; Enrolled in school: - Enrolled in grade 12"]].sort_values(
    by="Estimate; Enrolled in school: - Enrolled in grade 12",
    ascending = False
)
grade_12_ranked.head(25)

grade_12_ranked = grade_12_ranked.set_index("Neighborhood")

grade_12_ranked.head(25).plot(
    kind="bar",
    legend=False,
    figsize=(10,5),
    title="Top 25 Pittsburgh Neighborhoods by Grade 12 Enrollment"
)
```

Like the Kindergarten Enrollment dataset, the grade 12 dataset was arranged into a bar graph depicting the top 25 neighborhoods with highest student enrollment from the 12th grade. With Brookline being at #4, you can see how many students stay in the neighborhood throughout their years at grade school, suggesting that there is a strong family community and an abundance of accessible schools allowing this many students to enroll there. 

Overall, using kindergarten and grade 12 enrollment data allowed me to compare Pittsburgh neighborhoods based on two important stages of education. Even though some neighborhoods placed higher in individual categories, Brookline stood out for its strong and consistent rankings in both kindergarten and grade 12 enrollment. This suggests that Brookline attracts many families with young children and also retains students throughout high school. Based on this metric, Brookline is one of the best neighborhoods in Pittsburgh for overall K–12 educational engagement, and because of that I would give Brookline a rating of 4.5/5 for K-12 Student Enrollment.
