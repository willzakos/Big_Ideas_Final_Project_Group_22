Lets organize this info into graphs using pandas.


```python
import pandas as pd
facilities_list = pd.read_csv('facilities_list.csv', sep=",")
# cc_attendance_list.sample(10)
length=len(facilities_list['neighborhood'].value_counts(1).sort_index())
facilities_list['neighborhood'].value_counts(1).sort_index().iloc[0:length//2].plot(figsize=(20,6),kind='bar', y="Daily Attendance", rot=65)
#facilities_list['neighborhood'].value_counts(1).sort_index().iloc[length//2:length].plot(figsize=(20,6),kind='bar', y="Daily Attendance", rot=65)
# center_count=facilities_list['neighborhood'].value_counts(1).sort_index().plot(figsize=(20,6),kind='bar', y="Daily Attendance", rot=65)
# center_count.set_ylabel("Number Of Facilities Per Neighborhood")
# daily_attendance.set_ylim(0,100)
```

```python
facilities_list['neighborhood'].value_counts(1).sort_index().iloc[length//2:length].plot(figsize=(20,6),kind='bar', y="Daily Attendance", rot=65)
```

We can also see at a glance that Highland park has the most facilities at 0.07, or 7, and several neighborhoods (ex: Bon Air, Crafton Heights, etc) have the least amount at lower than 1 (representing bathrooms solely/no actual recreational centers). Specifically, Brookline is a happy middle ground in this graph, with about 5 facilities (counting hockey benches, pool, park, concession stands, and a shelter).
